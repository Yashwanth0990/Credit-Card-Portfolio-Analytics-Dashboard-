```sql
USE CATALOG credit_card_project;
USE SCHEMA gold;

-- ============================================================
-- Gold View 1: Customer 360
-- ============================================================

CREATE OR REPLACE VIEW credit_card_project.gold.vw_customer_360 AS
SELECT
    c.Customer_ID,
    c.First_Name,
    c.Last_Name,
    c.Age,
    c.Age_Group,
    c.Gender,
    c.City,
    c.City_Tier,
    c.State,
    c.Occupation,
    c.Annual_Income_Bracket,
    c.Credit_Score,
    c.Credit_Score_Band,
    c.Card_Type,
    c.Credit_Limit,
    c.Outstanding_Balance,
    c.Credit_Utilization_Ratio,
    c.Missed_Payments_Last_6M,
    c.Default_Status,
    c.Default_Label,
    c.Risk_Tier,
    c.Tenure_Years,
    c.Email,
    c.Phone
FROM credit_card_project.gold.dim_customer AS c;


-- ============================================================
-- Gold View 2: Transactions
-- ============================================================

CREATE OR REPLACE VIEW credit_card_project.gold.vw_transactions AS
SELECT
    f.Transaction_ID,
    f.customer_sk,
    f.merchant_sk,
    f.Date_ID,
    f.Transaction_Date,
    f.Transaction_Type,
    f.Transaction_Amount,
    f.Payment_Mode,
    f.Transaction_Status,
    f.Is_International,
    f.Card_Type,
    m.Merchant_Name,
    m.Category AS Merchant_Category,
    m.Is_Online,
    m.City AS Merchant_City
FROM credit_card_project.gold.fact_card_transactions AS f
LEFT JOIN credit_card_project.gold.dim_merchant AS m
    ON f.merchant_sk = m.merchant_sk;
```
